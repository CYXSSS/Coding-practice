```python
# load pandas, a data analysis package
# load numpy, a scientific computing package
# load scipy.stats, a module contains probability functions
# load matplotlib.pyplot, a framework provides a Matlab-like plotting
import pandas as pd
import numpy as np
import scipy.stats
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('Human height weight.xlsx',index_col = 'Index')
```

```python
m = data['Height'].mean()
print(f'The sample mean of height is {m}')
```

```python
v = data['Height'].var()
print(f'The sample variance of height is {v}')
```

```python
M = data['Height'].median()
print(f'The sample median of height is {M}')
```

```python
m1 = data['Height'].min()
print(f'The sample min height is {m1}')
```

```python
m2 = data['Height'].max()
print(f'The sample max height is {m2}')
```

```python
# create figure
fig, ax = plt.subplots()

# plot empirical height histogram
b = np.arange(55, 77 + 1, 0.5)
data['Height'].plot.hist(bins = b, density = True, ax = ax, label='Empirical Height')
```

```python
data2 = pd.read_excel('Rainfall2018July.xlsx',index_col = 'date')
# create figure
fig, ax = plt.subplots()
data2['rainfall'].plot.line( )
ax.set_title("Hong Kong Rainfall (mm) in July 2018")
ax.set_xlabel("Date")
ax.set_ylabel("Rainfall in mm")
```

```python
data3 = pd.read_excel('Carbatterylife.xlsx',index_col = 'Index')
# create figure
b = np.arange(1, 5 + 1, 0.5)
data3['Battery Life'].plot.hist(bins = b, density = True)
plt.title("Car Battery Life Histogram")
plt.ylabel("Relative Frequency")
```

```python
mb=data3['Battery Life'].mean()
print(mb)
```

```python
data3['Battery Life'].plot.box(title="Car Battery Life",grid=True)
```

```python
# create the plot
fig, ax = plt.subplots()

# scatter plot
ax.scatter(data['Height'], data['Weight'], alpha=0.3)
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid(True)
ax.set_title('The scatter plot')
```
